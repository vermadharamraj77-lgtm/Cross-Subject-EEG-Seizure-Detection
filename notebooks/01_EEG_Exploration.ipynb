```python
with open("../data/chb01-summary.txt", "r") as f:
    lines = f.readlines()

current_file = None

for line in lines:
    line = line.strip()

    if line.startswith("File Name:"):
        current_file = line.replace("File Name:", "").strip()

    if line.startswith("Number of Seizures in File:"):
        count = int(line.split(":")[-1].strip())

        if count > 0:
            print(current_file, "→", count, "seizure(s)")
```

```python
import mne

file_path = "../data/chb01_03.edf"

raw = mne.io.read_raw_edf(file_path, preload=True)

print("EEG loaded successfully!")
```

```python
print(raw)
```

```python
print("Sampling frequency:", raw.info["sfreq"], "Hz")
print("Number of channels:", len(raw.ch_names))
print("Duration:", raw.times[-1], "seconds")
```

```python
seizure = raw.copy().crop(
    tmin=2996,
    tmax=3036
)

print(seizure)
```

```python
seizure.plot(
    duration=20,
    n_channels=10,
    scalings="auto"
)
```

```python
print("Number of channels:", len(raw.ch_names))
```

```python
print(raw.ch_names)
```

```python
eeg_data = raw.get_data()

print(type(eeg_data))
print(eeg_data.shape)
```

```python
channel_1 = eeg_data[0]

print(channel_1.shape)
```

```python
print(channel_1[:20])
```

```python
sfreq = raw.info["sfreq"]

window_duration = 5

samples_per_window = int(sfreq * window_duration)

print("Samples in 5 seconds:", samples_per_window)
```

```python
segment = channel_1[:samples_per_window]

print(segment.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

time = raw.times[:samples_per_window]

plt.plot(time, segment)

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (µV)")
plt.title("5-Second EEG Segment")

plt.show()
```
